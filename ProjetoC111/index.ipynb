```python
import pandas as pd

df = pd.read_csv('movies.csv', on_bad_lines='skip', encoding='latin1', delimiter=',', encoding_errors='ignore')
```

```python
# Pergunta 3: Qual é a relação entre o orçamento e a receita — ou seja, quais filmes tiveram o maior ROI (retorno sobre investimento)?

df['budget'] = pd.to_numeric(df['budget'], errors='coerce')
df['revenue'] = pd.to_numeric(df['revenue'], errors='coerce')

df = df[df['revenue'] > 0]

df['roi'] = (df['revenue'] - df['budget']) / df['budget']

best_roi_movie = df.loc[df['roi'].idxmax()]

print("Movie with the best ROI:")
print(best_roi_movie[['title', 'budget', 'revenue', 'roi']])
```

```python
# Pergunta 4: Que gêneros tendem a gerar mais lucro com menor orçamento?

df['genres'] = df['genres'].astype(str).str.split(' ')
df = df.explode('genres')

genre_stats = df.groupby('genres').agg(
    average_roi=('roi', 'mean'),
    average_revenue=('revenue', 'mean'),
    average_budget=('budget', 'mean'),
    movie_count=('title', 'count')
).sort_values(by='average_roi', ascending=False)

print("Genres ranked by ROI (high revenue, low budget):")
print(genre_stats.head(10))
```

```python
#Pergunta 5: Quais os atores que aparecem mais de uma vez nos filmes mais bem avaliados?

df['vote_average'] = pd.to_numeric(df['vote_average'], errors='coerce')

well_evaluated_threshold = 8.5
high_rated_df = df[df['vote_average'] >= well_evaluated_threshold]

high_rated_df['cast'] = high_rated_df['cast'].astype(str).str.split('|')  # Adjust delimiter if needed
exploded_cast = high_rated_df.explode('cast')

actor_counts = exploded_cast['cast'].value_counts()

popular_actors = actor_counts[actor_counts > 1]

print("Actors who appeared more than once in highly rated movies:")
print(popular_actors)
```

```python
# Pergunta 6: Quais são os diretores com mais filmes lançados e qual é a média de avaliação (vote_average) dos seus filmes?

df['vote_average'] = pd.to_numeric(df['vote_average'], errors='coerce')

df = df[df['director'].notna()]

director_stats = df.groupby('director').agg(
    movie_count=('title', 'count'),
    average_rating=('vote_average', 'mean')
)

top_director = director_stats.sort_values(by='movie_count', ascending=False).head(1)

print("Director who launched the most movies:")
print(top_director)
```
